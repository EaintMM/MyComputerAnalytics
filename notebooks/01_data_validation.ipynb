```python
import pandas as pd

products = pd.read_csv("../data/raw/products.csv")
branches = pd.read_csv("../data/raw/branches.csv")
suppliers = pd.read_csv("../data/raw/suppliers.csv")
purchases = pd.read_csv("../data/raw/purchases.csv")
sales = pd.read_csv("../data/raw/sales.csv")
inventory = pd.read_csv("../data/raw/inventory.csv")
```

```python
print("Products:", products.shape)
print("Branches:", branches.shape)
print("Suppliers:", suppliers.shape)
print("Purchases:", purchases.shape)
print("Sales:", sales.shape)
print("Inventory:", inventory.shape)
```

```python
display(products.head())
display(branches.head())
display(suppliers.head())
display(purchases.head())
display(sales.head())
display(inventory.head())
```

### Checking missing values

```python
print("Missing values in products:")
print(products.isnull().sum())

print("\nMissing values in branches:")
print(branches.isnull().sum())

print("\nMissing values in suppliers:")
print(suppliers.isnull().sum())

print("\nMissing values in purchases:")
print(purchases.isnull().sum())

print("\nMissing values in sales:")
print(sales.isnull().sum())

print("\nMissing values in inventory:")
print(inventory.isnull().sum())
```

### Check duplicate

```python
print("Duplicate Product IDs:", products["ItemCode"].duplicated().sum())
print("Duplicate Purchase IDs:", purchases["PurchaseID"].duplicated().sum())
print("Duplicate Sale IDs:", sales["SaleID"].duplicated().sum())
```

### Check negative inventory

```python
print(
    "Negative inventory records:",
    (inventory["CurrentStock"] < 0).sum()
)
```

### Check inventory stock 

```python
inventory["CalculatedStock"] = (
    inventory["OpeningStock"]
    + inventory["PurchasedQty"]
    - inventory["SoldQty"]
)

inventory["StockCheck"] = (
    inventory["CalculatedStock"]
    == inventory["CurrentStock"]
)

print("Incorrect inventory calculations:")
print((inventory["StockCheck"] == False).sum())
```

### Check Sales calculation 

```python
sales["CalculatedRevenue"] = (
    sales["Quantity"] * sales["UnitSalePrice"]
)

sales["CalculatedCOGS"] = (
    sales["Quantity"] * sales["UnitCost"]
)

sales["CalculatedGrossProfit"] = (
    sales["CalculatedRevenue"] - sales["CalculatedCOGS"]
)

print(
    "Revenue errors:",
    (sales["Revenue"] != sales["CalculatedRevenue"]).sum()
)

print(
    "COGS errors:",
    (sales["COGS"] != sales["CalculatedCOGS"]).sum()
)

print(
    "Gross profit errors:",
    (sales["GrossProfit"] != sales["CalculatedGrossProfit"]).sum()
)
```

### Validation done. Clean temporary columns

```python
inventory.drop(
    columns=["CalculatedStock", "StockCheck"],
    inplace=True
)

sales.drop(
    columns=[
        "CalculatedRevenue",
        "CalculatedCOGS",
        "CalculatedGrossProfit"
    ],
    inplace=True
)
```

The following code is for updating products.csv to match with database schema.

```python
supplier_mapping = {
    "ABC Supplier": "SUP001",
    "XYZ Supplier": "SUP002",
    "Tech Supplier": "SUP003",
    "Local Supplier": "SUP004",
    "Network Supplier": "SUP005",
    "Software Supplier": "SUP006",
    "Computer Supplier": "SUP007"
}

products_db = products.copy()

products_db["supplier_code"] = products_db["Supplier"].map(supplier_mapping)

products_db = products_db.rename(columns={
    "ItemCode": "item_code",
    "ItemDescription": "item_description",
    "Category": "category",
    "BuyPrice": "buy_price",
    "SellPrice_LL": "sell_price_ll",
    "DealerPrice_LL": "dealer_price_ll"
})

products_db = products_db[
    [
        "item_code",
        "item_description",
        "category",
        "supplier_code",
        "buy_price",
        "sell_price_ll",
        "dealer_price_ll"
    ]
]

display(products_db.head())
```

```python
print("Shape:", products_db.shape)
print("Missing supplier codes:", products_db["supplier_code"].isnull().sum())
print("Supplier codes:", products_db["supplier_code"].unique())
```

```python
products_db.to_csv(
    "../data/processed/products_db.csv",
    index=False
)

print("Saved successfully.")
```
